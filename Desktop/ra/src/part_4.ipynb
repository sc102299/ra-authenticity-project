```python
import pandas as pd
import numpy as np
from pathlib import Path

DATA_DIR = Path("../data")

PART3_PATH = DATA_DIR / "part3_authenticity_index.csv"

part3 = pd.read_csv(PART3_PATH)

print(part3.shape)
part3.head()
```

```python
change_summary = (
    part3
    .groupby("changed_from_prior")["authenticity_index"]
    .agg(["count", "mean", "median", "min", "max"])
    .reset_index()
    .sort_values("mean", ascending=False)
)

change_summary
```

```python
changed_vs_stable = part3[
    part3["changed_from_prior"].isin(["changed", "not_changed"])
].copy()

changed_vs_stable_summary = (
    changed_vs_stable
    .groupby("changed_from_prior")["authenticity_index"]
    .agg(["count", "mean", "median", "min", "max"])
    .reset_index()
)

changed_vs_stable_summary
```

```python
change_by_sector = (
    changed_vs_stable
    .groupby(["sector", "changed_from_prior"])["authenticity_index"]
    .agg(["count", "mean", "median"])
    .reset_index()
    .sort_values(["sector", "changed_from_prior"])
)

change_by_sector
```

```python
changed_high = (
    part3[part3["changed_from_prior"] == "changed"]
    .sort_values("authenticity_index", ascending=False)
    .head(10)
)

changed_high[[
    "ticker",
    "company_name",
    "sector",
    "year",
    "authenticity_index",
    "theme_categories"
]]
```

```python
changed_low = (
    part3[part3["changed_from_prior"] == "changed"]
    .sort_values("authenticity_index", ascending=True)
    .head(10)
)

changed_low[[
    "ticker",
    "company_name",
    "sector",
    "year",
    "authenticity_index",
    "theme_categories"
]]
```

```python
def label_alignment(score):
    if pd.isna(score):
        return "missing"
    elif score >= 85:
        return "high_alignment"
    elif score >= 70:
        return "medium_alignment"
    else:
        return "low_alignment"

part3["alignment_group"] = part3["authenticity_index"].apply(label_alignment)

alignment_by_change = pd.crosstab(
    part3["changed_from_prior"],
    part3["alignment_group"],
    normalize="index"
)

alignment_by_change
```

```python
PART4_OUTPUT_PATH = DATA_DIR / "part4_change_alignment_summary.csv"

change_summary.to_csv(PART4_OUTPUT_PATH, index=False)

print(f"Saved Part 4 summary to: {PART4_OUTPUT_PATH}")
```
